```python
import pandas as pd
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt
import os
import math

# Configura tu archivo .xlsx
xlsx_path = r'C:\Users\akoun\Desktop\Biocruces\begibraintool\src\CEACO_image_selection.xlsx'
column_name = 'ceaco_relative_path'

# Lee el Excel
df = pd.read_excel(xlsx_path)
paths = df[column_name].drop_duplicates().tolist()

# Carga imágenes y nombres
images = []
filenames = []

for rel_path in paths:
    abs_path = os.path.abspath(rel_path)
    if os.path.exists(abs_path):
        img = Image.open(abs_path).convert("RGB")
        images.append(img)
        filenames.append(os.path.basename(rel_path))
    else:
        print(f"No encontrado: {abs_path}")

# Parámetros del mosaico
thumb_width = 200
thumb_height = 200
columns = 5
rows = math.ceil(len(images) / columns)
padding = 20
font_size = 16

# Fuente para el texto
try:
    font = ImageFont.truetype("arial.ttf", font_size)
except IOError:
    font = ImageFont.load_default()

# Tamaño de cada celda (imagen + texto debajo)
cell_height = thumb_height + font_size + 10
mosaic_width = columns * (thumb_width + padding)
mosaic_height = rows * (cell_height + padding)

# Crear el lienzo del mosaico
mosaic = Image.new('RGB', (mosaic_width, mosaic_height), color='white')
draw = ImageDraw.Draw(mosaic)

# Añadir imágenes al mosaico
for idx, (img, name) in enumerate(zip(images, filenames)):
    row = idx // columns
    col = idx % columns
    x = col * (thumb_width + padding)
    y = row * (cell_height + padding)

    # Redimensionar imagen
    img_resized = img.resize((thumb_width, thumb_height))
    mosaic.paste(img_resized, (x, y))

    # Dibujar nombre del archivo
    text_width, _ = draw.textsize(name, font=font)
    text_x = x + (thumb_width - text_width) // 2
    text_y = y + thumb_height + 5
    draw.text((text_x, text_y), name, fill='black', font=font)

# Mostrar el mosaico
mosaic.show()

# Guardar
mosaic.save('mosaico_ceaco.jpg')
```
