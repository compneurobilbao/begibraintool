```python
## UNIFICAR IMAGENES EN UNA CARPETA CON NOMBRE

path = r"C:\Users\akoun\Desktop\Biocruces\2.DATASETS\begibraintool_image_selection\02_ANIMALES_256_ObjectCategories_SEGMENTADOS"

import os
import shutil

def extract_silhouette_images(path, exts=None, dry_run=True):
    """
    Busca dentro de 'path' imágenes en subcarpetas llamadas 'silhouette' y las mueve a 'path',
    renombrándolas con los nombres de las carpetas superiores.
    El resto de subcarpetas se eliminan al final.
    
    exts: lista de extensiones a considerar (por ejemplo ['.jpg','.png'])
    dry_run: True para solo mostrar acciones, False para ejecutarlas
    """
    if exts:
        exts = [e.lower() for e in exts]

    for root, dirs, files in os.walk(path):
        # Comprobar si estamos en una carpeta llamada 'silhouette'
        if os.path.basename(root).lower() == 'silhouette':
            # Determinar nombre combinado de carpetas superiores
            rel_path = os.path.relpath(root, path)  # e.g. Cat1\Sub2\Sub3\silhouette
            parts = rel_path.split(os.sep)[:-1]      # quitamos 'silhouette'
            prefix = "_".join(parts)                # e.g. Cat1_Sub2_Sub3

            for filename in files:
                src = os.path.join(root, filename)
                if os.path.isfile(src):
                    if exts and os.path.splitext(filename)[1].lower() not in exts:
                        continue
                    new_filename = f"{prefix}_{filename}"
                    dst = os.path.join(path, new_filename)

                    # Evitar sobrescrituras
                    base, ext = os.path.splitext(new_filename)
                    counter = 1
                    while os.path.exists(dst):
                        new_filename = f"{base}_{counter}{ext}"
                        dst = os.path.join(path, new_filename)
                        counter += 1

                    if dry_run:
                        print(f"[MOVE] {src} -> {dst}")
                    else:
                        shutil.move(src, dst)

    # Eliminar carpetas residuales
    for folder in os.listdir(path):
        folder_path = os.path.join(path, folder)
        if os.path.isdir(folder_path):
            if dry_run:
                print(f"[DELETE] {folder_path}")
            else:
                shutil.rmtree(folder_path)
# Ejemplo de uso:
extract_silhouette_images(path, exts=['.jpg','.png'], dry_run=False)
```
